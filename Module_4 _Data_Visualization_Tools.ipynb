#STEP 2 — Combine Both Datasets
Combine Chicago and New Orleans datasets
We merge both datasets into one single dataframe called combined. This unified dataset will be used for cleaning and dashboard building.



```python
import pandas as pd

chicago = pd.read_csv("listings (1).csv")
new_orleans = pd.read_csv("listings.csv")

print("Chicago Loaded:", chicago.shape)
print("New Orleans Loaded:", new_orleans.shape)
```

```python
# STEP 2.1 — Add City Column
chicago["city"] = "Chicago"
new_orleans["city"] = "New Orleans"

# STEP 2.2 — Combine both datasets
combined = pd.concat([chicago, new_orleans], ignore_index=True)

print("Combined Shape:", combined.shape)
combined.head()
```

#STEP 3 — Clean and Standardize Column Names
We clean column names to make them easier to work with. This step will:

Convert all column names to lowercase

Remove extra spaces

Replace spaces with underscores _

Make all columns consistent and compatible with Tableau and Python

Clean column names help avoid future errors during analysis and visualisation.

```python
# STEP 3 — Clean column names
combined.columns = (
    combined.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
)

combined.columns
```

#STEP 4 — Handle Missing Values

In this step, we will check which columns have missing values.
Missing values can cause errors in Tableau and affect visualizations.

We will:

Check missing values for each column

Drop rows with too many missing values

Fill missing values for important columns

Ensure the dataset is clean and ready for analysis

```python
# STEP 4 — Check Missing Values
missing = combined.isnull().sum()
print("Missing Values in Each Column:")
missing
```

```python
# STEP 4.1 — Remove rows with too many missing values
combined = combined.dropna(thresh=5)

# STEP 4.2 — Fill missing categorical columns with "Unknown"
combined['neighbourhood_group'] = combined['neighbourhood_group'].fillna("Unknown")
combined['last_review'] = combined['last_review'].fillna("2000-01-01")

# STEP 4.3 — Fill missing numerical columns with 0
combined['reviews_per_month'] = combined['reviews_per_month'].fillna(0)

print("Missing values cleaned!")
combined.isnull().sum()
```

#STEP 5 — We convert important columns into correct data types.

* last_review → convert to datetime

* Price columns sometimes import as text → convert to numeric

* Review-related columns must be float/int

* city works best as category

* IDs should be int

* Makes dataset Tableau-ready with correct data types




```python
# STEP 5 — Convert Data Types

# Convert last_review to datetime
combined['last_review'] = pd.to_datetime(combined['last_review'], errors='coerce')

# Convert price to numeric (remove symbols if any)
combined['price'] = pd.to_numeric(combined['price'], errors='coerce')

# Convert reviews_per_month to float
combined['reviews_per_month'] = pd.to_numeric(combined['reviews_per_month'], errors='coerce')

# Convert id and host_id to integers
combined['id'] = combined['id'].astype('int64')
combined['host_id'] = combined['host_id'].astype('int64')

# Convert availability_365 to int
combined['availability_365'] = combined['availability_365'].astype('int64')

# Convert city to category (better for Tableau & memory)
combined['city'] = combined['city'].astype('category')

print("STEP 5 Completed — Data types converted successfully!")
combined.dtypes
```

#STEP 6 — Create New Feature Columns

In this step, we will create three new columns that will help us perform better analysis and build more meaningful dashboards in Tableau.

##Why create new feature columns?

These additional columns help in:

* Better segmentation

* Easy filtering

* Improved insights

* Cleaner visualizations




#Step 6.1 — Create price_category Column

We create a new column that categorizes each listing based on its price:

* Low → Price less than 100

* Medium → Price between 100 and 250

* High → Price greater than 250

This helps in comparing property prices across cities.



#Step 6.2 — Create is_super_host Column

Hosts with a higher number of listings are usually more experienced.

We classify hosts as:

* Yes → If the host has more than 10 listings

* No → If the host has 10 or fewer listings

This will help in analyzing the difference between normal and super hosts.

#Step 6.3 — Create days_since_last_review Column

To understand how active a listing is, we calculate the number of days since its last review.

Formula:

```python
days_since_last_review = today’s date – last review date
```

If a listing has no review date, we set the value to 0.

This column helps in identifying recent and active listings.

```python
# STEP 6 — Create New Feature Columns

import numpy as np
import datetime as dt

# 6.1 - Create price_category
combined["price_category"] = pd.cut(
    combined["price"],
    bins=[0, 100, 250, combined["price"].max()],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

# 6.2 - Create is_super_host (Based on listings count)
combined["is_super_host"] = np.where(
    combined["calculated_host_listings_count"] > 10,
    "Yes",
    "No"
)

# 6.3 - Create days_since_last_review
today = pd.to_datetime("2025-11-22")  # You can set today’s date

combined["days_since_last_review"] = (
    today - combined["last_review"]
).dt.days

# Replace negative or NaN values with 0
combined["days_since_last_review"] = combined["days_since_last_review"].fillna(0)
combined["days_since_last_review"] = combined["days_since_last_review"].clip(lower=0)

# Show new columns
combined[["price", "price_category", "calculated_host_listings_count", "is_super_host", "last_review", "days_since_last_review"]].head()
```
